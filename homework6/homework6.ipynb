```python
arr = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])
# ---Brainstorm---
#need funciton to check a # is prime and then another one to check the number against the array
# 2 is the only even prime number.
# need to check for odd divisors from 3 up to the square root of n.
# if no divisors are found, its prime.
def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
    return True

# 1. Create a vectorized version of is_prime function to let NumPy  apply the Python function to every single element in the array efficiently.
# 2. apply the vectorized function to the entire array which creates a new array of the same shape with bool values.
# 3. then check axis = 1 (rpw) for at least one true value and collapse the 2D mask into a 1D mask for the rows.
# 4. Use this 1D boolean mask to select rows from the original array.

def find_prime_rows(arr):
    
    
    vectorized_is_prime = np.vectorize(is_prime)

    prime_mask = vectorized_is_prime(arr)

    row_has_prime = np.any(prime_mask, axis=1)

    filtered_array = arr[row_has_prime]
    
    return filtered_array

prime_rows = find_prime_rows(arr)
print(prime_rows)
print(arr)
        
```

```python
def make_board():
    checkerboard = np.zeros((8, 8), dtype = int)
    checkerboard[::2, 1::2] = 1 #even rows and odd columns get a 1
    checkerboard[1::2, ::2] = 1 #even columns and odd rows get a 1
    return checkerboard

checkerboard = make_board()
print(checkerboard)
```

```python
def reverse_board():
    checkerboard = np.zeros((8, 8), dtype = int)
    checkerboard[::2, ::2] = 1
    checkerboard[1::2, 1::2] = 1
    return checkerboard

checkerboard = reverse_board()
print(checkerboard)
```

```python
def expand(s, num_spaces):
    spacer = " " * num_spaces #spacer literally multiplies spacebar by input #
    
    return spacer.join(s) # join every character in the string w/ the spacer

# Create the main 'expansion' function by vectorizing the helper which tells numpy to apply the expand function
# to every element in the input array.
expansion = np.vectorize(expand)
string = np.array(['where','are','you','going?'])
expanded_1 = expansion(string, 1)
print(expanded_1)
expanded_10 = expansion(string, 10)
print(expanded_10)
```

```python
import numpy as np

def secondBrightest(stars_array):
  sorted_stars = np.sort(stars_array, axis=1)
  return sorted_stars[:, -2]

np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))

print(stars)

result = secondBrightest(stars)

print(result)
```
